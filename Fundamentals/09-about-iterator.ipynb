```python
from itertools import groupby

data=[
    {'dept':'IT','amount':350},
    {'dept':'Audit','amount':200}, 
    {'dept':'HR','amount':300},
    {'dept':'Audit','amount':700},
    {'dept':'IT','amount':1000},
    {'dept':'HR','amount':600}
]
```

```python
sorted_data =sorted(data, key=lambda x:x['dept'])
```

```python
sorted_data
```

```python
new_grp=groupby(sorted_data,key=lambda x:x['dept'])
```

```python
for dept, group in groupby(sorted_data,key=lambda x:x['dept']):
    total=sum(i['amount'] for i in group)
    print(f'Total expenditure for {dept} is {total}')
```

```python
from itertools import count

for num in count(start=1, step=2):
    print(num)
    if num==5:
        break
```

```python
from itertools import cycle

rainbow=['Violet','Blue','Green','Yellow','Orange','Red']

count=0
for color in cycle(rainbow):
    print(color)
    count+=1
    if count==10:
        break
```

```python
from itertools import repeat

for msg in repeat('Loading ...',3):
    print(msg)
```

```python
from itertools import permutations

players=['RCB','KK','CSK','MI']

for p in permutations(players,2):
    print(p)
```

```python
from itertools import combinations

players=['RCB','KK','CSK','MI']

for p in combinations(players,2):
    print(p)
```

```python
from itertools import islice

data=iter(range(30))
for i in islice(data,7,11): # similar to slicing - data[7:11]
    print(i)
```

```python
from itertools import tee

data=iter(range(1,8))
a,b=tee(data,2)

for i in a:
    print(i)
```

```python
from itertools import product
starters=['soup','salad']
main=['pasta','Biryani']
dessert=['Icecream','Brownie']

for combo in product(starters,main,dessert):
    print(combo)
```
