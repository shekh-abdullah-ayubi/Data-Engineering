### __Constructor__
In Python, a `constructor` is a special method used to initialize the attributes of an object when a class is instantiated. It is automatically called whenever a `new object` of the class is created.\
Key characteristics of Python constructors:
* __`___init__` method__: The constructor in Python is defined using the special method `__init__`(self, ...). The self parameter refers to the instance of the object being created, and it is the first parameter in the `__init__` method.
* __Automatic invocation__: Unlike regular methods that need to be explicitly called, the `__init__` method is automatically invoked when you create an object of a class.
* __Initialization of attributes__: The primary purpose of the constructor is to initialize the instance variables (attributes) of the object with initial values, either default values or values passed as arguments during object creation.
* __Parameters__: Constructors can accept parameters, allowing you to pass initial values to the object's attributes when creating an instance.

_Note :_ If you don't define an `__init__` method in your class, <u>Python implicitly provides a default constructor with an empty body.</u> 

```python
### Instance Variable and Methods
class Dog:
    ## constructor
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
dog1= Dog('Bruno',13)
f'{dog1.name} ages is {dog1.age} yrs.'
```

```python
dog2=Dog('Sheru',12)
f'{dog2.name} is {dog2.age} yrs. older'
```

```python
## Define a class with instance methods
class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def bark(self):
        print(f'{self.name} says Smoking Kills ...')

dog1=Dog('Bruno',13)
dog1.bark()
dog2=Dog('Entertainment',14)
dog2.bark()
```

Lets Model a Bank Account

```python
## Define a class for bank account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.bal=balance
    def deposit(self,amount):
        self.bal+=amount
        print(f'{amount} is deposited. Updated balance is {self.bal}')
    def withdraw(self,amount):
        if amount>self.bal:
            print('Insufficient Fund')
        else:
            self.bal-=amount
            print(f'{amount} is withdrawn. New Balance is {self.bal}')
    def credit(self,score,amount):
        if score<6:
            print('Inappropriate CBIL score')
        else:
            self.bal+=amount
            print(f'credit amount: {amount} is auto-approved !,\nBalance after credit {self.bal}')
    def get_balance(self):
        return self.bal

account=BankAccount('Pushpa Raj',70000)
f'A/C Balance info- Customer Name: {account.owner}, Balance: {account.bal}'
```

```python
account.deposit(100000)
```

```python
account.withdraw(50000)
```

```python
account.credit(5,80000)
```

```python
account.credit(7,80000)
```

```python
account.get_balance()
```
